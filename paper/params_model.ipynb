```python
import sys
import os
sys.path.append(os.path.abspath('../code'))

from models.xresnet1d import xresnet1d101
from models.resnet1d import resnet1d_wang, wrn1d_22
from models.inception1d import inception1d
from models.rnn1d import RNN1d
```

```python
def count_params(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

jumlah_channel = 5
jumlah_class = 5

xresnet = xresnet1d101(input_channels=jumlah_channel, num_classes=jumlah_class)
resnetwang = resnet1d_wang(input_channels=jumlah_channel, num_classes=jumlah_class)
wrn = wrn1d_22(input_channels=jumlah_channel, num_classes=jumlah_class)
inception = inception1d(input_channels=jumlah_channel, num_classes=jumlah_class)
rnn = RNN1d(input_channels=jumlah_channel, num_classes=jumlah_class)

print("Parameter XResNet    :", count_params(xresnet))
print("Parameter ResNet Wang:", count_params(resnetwang))
print("Parameter Wide ResNet:", count_params(wrn))
print("Parameter Inception  :", count_params(inception))
print("Parameter RNN        :", count_params(rnn))
```
